```python
from semanticscholar import SemanticScholar
import pandas as pd
```

```python
# Initialize Semantic Scholar API
sch = SemanticScholar()
```

```python
# Load the papers to be snowballed
paper_roots = pd.read_csv('data/cleaned/edu/search-results-master-v3-full.csv')
```

# Forward Search -- get all references of the papers in a csv file

```python
# Prepare DataFrame to store the results
columns = [
    'ReferencedPaperTitle',
    'ReferencedDOI',
    'ReferencedURL',
    'FromDOI',
]
results_df = pd.DataFrame(columns=columns)

j = 0
for i in range(len(paper_roots)):
    paper = paper_roots.iloc[i]
    paper_doi = paper["DOI"]
    print(paper_doi)
    if paper_doi == '' or paper_doi is None or pd.isna(paper_doi):
        print(f"WARNING: paper with title {paper["PaperTitle"]} has no DOI, please update DOI or proceed manually")
        continue
    sch_paper = sch.get_paper(paper_doi)
    references = sch_paper['references']
    if references is None:
        continue
    for referenced_paper in references:
        # do the same process for each referenced paper (same as what scraper does)
        # but for now, just get the paper name and doi
        if referenced_paper['externalIds'] is None:
            print("    No externalIds")
            continue
        referenced_paper_doi = referenced_paper['externalIds'].get('DOI', '')
        if referenced_paper_doi == '':
            print("    No DOI")
        else:
            print("   ", referenced_paper_doi)
        new_paper = {
            'ReferencedPaperTitle': referenced_paper['title'],
            'ReferencedDOI': referenced_paper_doi,
            'ReferencedURL': referenced_paper['url'],
            'FromDOI': paper_doi,
        }
        results_df = pd.concat([results_df, pd.DataFrame([new_paper])], ignore_index=True)
    
    # TODO: remove the following line
    j += 1
    if j == 10:  
        break
results_df.to_csv('data/forward-snowballed-result.csv', index=False)
```

# Backward Search -- get all papers that cite the papers in a csv file

```python
# Prepare DataFrame to store the results
columns = [
    'PaperCitedTitle',
    'PaperCitedDOI',
    'PaperCitedURL',
    'FromDOI',
]
results_df = pd.DataFrame(columns=columns)

j = 0
for i in range(len(paper_roots)):
    paper = paper_roots.iloc[i]
    paper_doi = paper["DOI"]
    print(paper_doi)
    if paper_doi == '' or paper_doi is None or pd.isna(paper_doi):
        print(f"WARNING: paper with title {paper["PaperTitle"]} has no DOI, please update DOI or proceed manually")
        continue
    sch_paper = sch.get_paper(paper_doi)
    if sch_paper['citations'] is None or sch_paper['citationCount'] == 0:
        print(f"   paper with title {sch_paper['title']} has no citations")
        continue
    references = sch_paper['citations']
    for referenced_paper in references:
        # do the same process for each referenced paper (same as what scraper does)
        # but for now, just get the paper name and doi
        if referenced_paper['externalIds'] is None:
            print("    No externalIds")
            continue
        referenced_paper_doi = referenced_paper['externalIds'].get('DOI', '')
        if referenced_paper_doi == '':
            print("    No DOI")
        else:
            print("   ", referenced_paper_doi)
        new_paper = {
            'PaperCitedTitle': referenced_paper['title'],
            'PaperCitedDOI': referenced_paper_doi,
            'PaperCitedURL': referenced_paper['url'],
            'FromDOI': paper_doi,
        }
        results_df = pd.concat([results_df, pd.DataFrame([new_paper])], ignore_index=True)
        
    # TODO: remove the following line
    j += 1
    if j == 10:  
        break
results_df.to_csv('data/backward-snowballed-result.csv', index=False)
```
