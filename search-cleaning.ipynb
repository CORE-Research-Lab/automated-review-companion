```python
import pandas as pd
```

```python
edu_master_df = pd.read_csv('data/cleaned/edu/search-results-master.csv')
social_master_df = pd.read_csv('data/cleaned/social/search-results-master.csv')
```

```python
# iterate through each row in the master df and remove the row if "PaperTitle" contain the following keywords
KEYWORDS = ["oil", "tree", "forest", "agriculture", "soil", "palms", "bio"]
edu_master_df_cleaned = edu_master_df[~edu_master_df['PaperTitle'].str.lower().str.contains('|'.join(KEYWORDS), case=False, regex=True)]
social_master_df_cleaned = social_master_df[~social_master_df['PaperTitle'].str.lower().str.contains('|'.join(KEYWORDS), case=False, regex=True)]
```

```python
edu_master_df_cleaned.to_csv('data/cleaned/edu/search-results-master-v2.csv', index=False)
social_master_df_cleaned.to_csv('data/cleaned/social/search-results-master-v2.csv', index=False)
```
